```python
from langchain.chat_models import init_chat_model
from langchain_groq import ChatGroq
import os
from dotenv import load_dotenv
load_dotenv()
api_key = os.environ.get("GROQ_API_KEY")
model   = ChatGroq(
            model="qwen/qwen3-32b",
            temperature=0,
            max_tokens=None,
            reasoning_format="parsed",
            timeout=None,
            max_retries=2,
            # other params...
    )
```

```python
from langchain.tools import tool
@tool
def tool1(x:int)->int:
    """this tool returns the 24th power of the input"""
    return x**24

#we can overwrite it to return the 25th
@tool("tool1")
def tool2(x:int)->int:
    """this tool returns the 25th power of the input"""
    return x**25
from tavily import TavilyClient
from typing import Any
tavelyclient = TavilyClient(api_key=os.environ.get("TAVELY_API_KEY"))
@tool
def websearchtool(query:str)->str:
    """this tool performs a web search and returns the result"""
    
    return tavelyclient.search(query)

```

```python
from langchain.agents import create_agent
messages = [
    {"role":"system", "content":"You are a helpful assistant."},
    {"role":"user", "content":"What is the capital of France?"}
]
agent = create_agent(model = model,
                     system_prompt = "You are a helpful assistant.",
                     tools = [tool1,tool2,websearchtool]
                     
                     )
response = agent.invoke({"messages": messages})
messages.append({"role":"assistant", "content": response['messages'][-1].content})
print(response['messages'][-1].content)

```

```python
def user_query_and_get_response(query,messages):
    messages.append({"role":"user", "content": query})
    response = agent.invoke({"messages": messages})
    messages.append({"role":"assistant", "content": response['messages'][-1].content})
    # print(response)
    print(response['messages'][-1].content)
    
```

```python
user_query_and_get_response("who won in 2026 west bengal elections?",messages)
```
